```python

import numpy as np

# Coordinates of 8 cities
cities = np.array([
    [0, 0], [2, 6], [5, 3], [6, 7],
    [8, 2], [1, 9], [9, 6], [4, 8]
])

n = len(cities)

# Calculate distance between cities
distance = np.linalg.norm(
    cities[:, None] - cities[None, :], axis=2
)

# Initialize pheromones
pheromone = np.ones((n, n))

# ACO parameters
ants = 20
iterations = 100
alpha = 1
beta = 2
evaporation = 0.5
Q = 100

best_route = None
best_distance = float('inf')

np.random.seed(42)
rng = np.random.default_rng(42)

# ACO algorithm
for iteration in range(iterations):

    all_routes = []

    for ant in range(ants):

        route = [0]
        unvisited = set(range(1, n))

        while unvisited:
            current = route[-1]
            choices = np.array(sorted(unvisited))

            probability = (
                pheromone[current, choices] ** alpha
                * (1 / (distance[current, choices] + 1e-10)) ** beta
            )

            probability = probability / probability.sum()

            next_city = int(rng.choice(choices, p=probability))

            route.append(next_city)
            unvisited.remove(next_city)

        # Calculate total route distance
        total = sum(
            distance[route[i], route[i + 1]]
            for i in range(n - 1)
        )
        total += distance[route[-1], route[0]]

        all_routes.append((route, total))

        if total < best_distance:
            best_distance = total
            best_route = route.copy()

    # Pheromone evaporation
    pheromone *= (1 - evaporation)

    # Pheromone update
    for route, total in all_routes:
        for i in range(n):
            a = route[i]
            b = route[(i + 1) % n]

            pheromone[a, b] += Q / total
            pheromone[b, a] += Q / total

# Display result
print("Best Delivery Route:")
print(" -> ".join(map(str, best_route + [best_route[0]])))

print("Shortest Distance:", round(best_distance, 2))
```
