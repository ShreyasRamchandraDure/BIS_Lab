```python

import numpy as np
import math

# Objective function
def fitness(x):
    return x**2 + 4*x + 5

# Parameters
n = 15
iterations = 100
pa = 0.25

lower = -10
upper = 10

np.random.seed(42)
rng = np.random.default_rng(42)

# Initialize nests
nests = rng.uniform(lower, upper, n)

# Find initial best nest
best = float(nests[np.argmin([fitness(x) for x in nests])])
best_fitness = fitness(best)

# Levy flight parameters
beta = 1.5

sigma = (
    math.gamma(1 + beta) *
    math.sin(math.pi * beta / 2) /
    (math.gamma((1 + beta) / 2) *
     beta * 2**((beta - 1) / 2))
)**(1 / beta)

# Cuckoo Search algorithm
for iteration in range(iterations):

    # Generate new solutions using Levy flights
    for i in range(n):

        u = rng.normal(0, sigma)
        v = rng.normal(0, 1)

        levy = u / (abs(v)**(1 / beta))

        new_solution = nests[i] + 0.01 * levy * (nests[i] - best)

        new_solution = np.clip(new_solution, lower, upper)

        # Select a random nest
        j = int(rng.integers(n))

        # Replace nest if better solution is found
        if fitness(new_solution) < fitness(nests[j]):
            nests[j] = new_solution

    # Abandon worst nests
    abandon = rng.random(n) < pa
    nests[abandon] = rng.uniform(lower, upper, abandon.sum())

    # Update best solution
    current_best = nests[np.argmin([fitness(x) for x in nests])]

    if fitness(current_best) < best_fitness:
        best = float(current_best)
        best_fitness = fitness(best)

# Display result
print("Best Solution (x):", round(best, 4))
print("Minimum Function Value:", round(best_fitness, 4))
```
